# Simple RNN with Keras

## Importing the libraries

```python
import numpy as np
import tensorflow as tf
```

## Creating documents with labels

```python
documents = [
    "Deep learning uses neural networks.",
    "LSTMs are powerful sequence models.",
    "I enjoy studying machine learning.",
    "Deep learning is truly amazing.",
    "Natural language processing is fascinating.",
    "Poorly trained models produce inaccurate results.",
]
```

```python
labels = np.array([1, 1, 1, 0, 0, 0])
```

```python
documents, labels
```

## Initializing TextVectorization

```python
from tensorflow.keras.layers import TextVectorization
```

```python
vectorizer = TextVectorization(
    max_tokens = 15000,
    standardize = 'lower_and_strip_punctuation',
    split = 'whitespace',
    output_mode = 'int',
    output_sequence_length = 8
)
```

```python
vectorizer.adapt(documents)
```

```python
encoded = vectorizer(documents)
```

## Creating a Simple RNN Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
model = Sequential()

model.add(
    Embedding(
        input_dim=len(vectorizer.get_vocabulary()) + 1,
        output_dim = 8,
        input_length = 8
    )
)

model.add(SimpleRNN(10))
model.add(Dense(1, activation='sigmoid'))
```

## Compiling the model

```python
model.compile(
    optimizer = 'adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

## Training the model

```python
model.fit(
    encoded,
    labels,
    epochs = 10,
    verbose = 1,
    validation_split = 0.2
)
```
