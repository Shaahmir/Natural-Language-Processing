# TF-IDF Implementation

## Importing the libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Creating a Dataframe

```python
df = pd.DataFrame({'text':["NLP processes text data.", "Machine learning processes data.", "Deep learning processes text and data."], 'label':[0, 1, 1]})
```

```python
df
```

## Importing TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

## Initializing TF-IDF Vectorizer

```python
tfidf = TfidfVectorizer()
```

```python
X = tfidf.fit_transform(df['text'])
```

```python
X.toarray().round(2)
```

## Printing the values

```python
print(tfidf.idf_.round(2))
```

```python
print(tfidf.get_feature_names_out())
```

## Visualizing as Heatmap

```python
df_tfidf = pd.DataFrame(
    X.toarray(),
    columns = tfidf.get_feature_names_out()
)
```

```python
plt.figure(figsize = (10, 4))
sns.heatmap(df_tfidf, annot = True)
plt.title("TF-IDF Scores Across Document")
plt.xlabel("Terms")
plt.ylabel("Document")
plt.show()
```
