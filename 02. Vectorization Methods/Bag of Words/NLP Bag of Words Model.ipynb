# Implementing Bag of Words Model

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

## Creating documents

```python
documents =  [
    "NLP processes text data.",
    "Machine learning processes data.",
    "Deep learning processes text and data."
]
```

## Importing CounterVectorizer Class

```python
from sklearn.feature_extraction.text import CountVectorizer
```

## Initializing CounterVectorizer

### &nbsp; Initialize CountVectorizer with multiple parameters:

*   `ngram_range = (1, 2)` &nbsp; -> &nbsp; **Use unigrams and bigrams**
*   `max_df = 0.85` &nbsp; -> &nbsp; **Ignore words that appear in more than 85% of the documents**
*   `min_df = 2` &nbsp; -> &nbsp; **Ignore words that appear in fewer than 2 documents**
*   `max_features = 5` &nbsp; -> &nbsp; **Only keep the top 5 features**

```python
cv = CountVectorizer(stop_words = 'english')
```

```python
X = cv.fit_transform(documents)
```

## Converting it into array

```python
X.toarray()
```

## Printing Vocabulary with Indexes

```python
print("\nVocabulary with index")
print(cv.vocabulary_)
```

## Printing Vocabulary

```python
print("\nFeature Names (Vocabulary):")
print(cv.get_feature_names_out())
```

## Graphically Visualizing BoW

```python
words = cv.get_feature_names_out()
words_count = np.sum(X.toarray(), axis = 0)
```

```python
plt.bar(words, words_count)
plt.title("Word Frequencies in document")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```

## Visualizing as Heatmap

```python
df_bow = pd.DataFrame(
    X.toarray(),
    columns = cv.get_feature_names_out()
)
```

```python
sns.heatmap(df_bow, annot = True)
plt.title("BoW Scores Across Document")
plt.xlabel("Terms")
plt.ylabel("Document")
plt.show()
```
