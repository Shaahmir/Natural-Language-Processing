# Implementing N-Grams

## Importing the libraries

```python
import pandas as pd
import numpy as np
```

## Creating a dataframe

```python
df = pd.DataFrame({'text':["NLP processes text data.", "Machine learning processes data.", "Deep learning processes text and data."], 'label':[0, 1, 1]})
```

```python
df
```

## Importing CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer
```

## Uni-Gram

```python
cv = CountVectorizer(ngram_range = (1, 1))
```

```python
X = cv.fit_transform(df['text'])
```

```python
print("Unigrams:\n")
print(cv.get_feature_names_out())
```

```python
X.toarray()
```

## Bi-Gram

```python
cv = CountVectorizer(ngram_range = (2, 2))
```

```python
X = cv.fit_transform(df['text'])
```

```python
print("Bigrams:\n")
print(cv.get_feature_names_out())
```

```python
X.toarray()
```

## Tri Gram

```python
cv = CountVectorizer(ngram_range = (3, 3))
```

```python
X = cv.fit_transform(df['text'])
```

```python
print("Trigrams:\n")
print(cv.get_feature_names_out())
```

```python
X.toarray()
```
