# Implementing Lemmetization with NLTK

## Importing the libraries

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download('wordnet')
```

## Creating documents

```python
documents = [
    "The cats are running and jumping in the garden.",
    "She is a beautiful runner and loves to run fast.",
    "Running helps to build stamina and strength.",
    "He ran swiftly and caught the ball."
]
```

## Initializing Lematization Class

```python
lemmatizer = WordNetLemmatizer()
```

## Creating Function to Apply Lematization

```python
def apply_nltk_lemmatization(doc):

    # Tokenizing the document
    tokens = word_tokenize(doc)

    # Applying lemmatization to each token
    lemmatized_tokens = [lemmatizer.lemmatize(token,pos='v') for token in tokens]

    return {
        "original": tokens,
        "lemmatized": lemmatized_tokens
    }
```

## Printing the result

```python
for doc in documents:
    results = apply_nltk_lemmatization(doc)
    print(f"Original Document: {results['original']}")
    print(f"Lemmatized: {results['lemmatized']}")
    print("-" * 40)
```

# Implementing Lemmetization with SpaCy

## Importing the libraries

```python
import spacy
```

## Initializing Lematization Class

```python
nlp = spacy.load("en_core_web_sm")
```

## Creating Function to Apply Lemmatization

```python
def apply_spacy_lemmatization(doc):

    # Tokenizing the document
    tokens = nlp(doc)

    # Applying lemmatization to each token
    original_tokens = [token.text for token in tokens]
    lemmatized_tokens = [token.lemma_ for token in tokens]

    return {
        "original": original_tokens,
        "lemmatized": lemmatized_tokens
    }
```

## Printing the result

```python
for doc in documents:
    results = apply_spacy_lemmatization(doc)
    print(f"Original Document: {results['original']}")
    print(f"Lemmatized: {results['lemmatized']}")
    print("-" * 40)
```
